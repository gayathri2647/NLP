```python
# Sample Corpora
corpus = [
    "<s> she likes tea </s>",
    "<s> she likes coffee </s>",
    "<s> tea is hot </s>"
]
```

```python
# Tokenization
tokens_list = [s.lower().split() for s in corpus]
```

```python
# Count unigrams and bigrams
unigrams = {}
bigrams = {}

for tokens in tokens_list:
    for i in range(len(tokens)):
        word = tokens[i]

        # Unigram count (single word)
        unigrams[word] = unigrams.get(word, 0) + 1

        # Bigram count (two-word pair)
        if i < len(tokens) - 1:
            pair = (tokens[i], tokens[i+1])
            bigrams[pair] = bigrams.get(pair, 0) + 1
```

```python
# Display counts
print("Individual Word Count (Unigrams):")
for k, v in unigrams.items():
    print(k, ":", v)

print("\nTwo-Word Occurrence Count (Bigrams):")
for k, v in bigrams.items():
    print(k, ":", v)
```

```python
# Bigram probabilities
bigram_prob = {}
for (w1, w2), count in bigrams.items():
    bigram_prob[(w1, w2)] = count / unigrams[w1]
```

```python
# NEW sentence (not in corpus)
new_sentence = ["<s>", "she", "likes", "tea", "is", "hot", "</s>"]
```

```python
# Probability calculation using formula
prob = 1

for i in range(len(new_sentence) - 1):
    pair = (new_sentence[i], new_sentence[i+1])

    if pair in bigram_prob:
        prob *= bigram_prob[pair]
    else:
        prob = 0
        break
```

```python
# Output
print("\nProbability of New Sentence:")
print(prob)
```
