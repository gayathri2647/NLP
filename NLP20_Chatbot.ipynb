```python
# Import library
import random
```

```python
# Sample data
greetings = ["hello", "hi", "hey"]
farewells = ["bye", "goodbye"]
```

```python
responses = {
    "hello": ["Hello!", "Hi there!", "Hey!"],
    "hi": ["Hello!", "Hi there!", "Hey!"],
    "hey": ["Hello!", "Hi there!", "Hey!"],
    "bye": ["Goodbye!", "See you later!"],
    "goodbye": ["Goodbye!", "See you later!"]
}
```

```python
questions = {
    "your name": "I'm a simple chatbot.",
    "how are you": "I'm doing fine, thank you!",
    "what can you do": "I can chat with you.",
    "help": "You can say hello, ask questions, or say bye."
}
```

```python
while True:
    user = input("You: ").lower()
    # Greeting
    if user in greetings:
        print("Chatbot:", random.choice(responses[user]))
    # Farewell
    elif user in farewells:
        print("Chatbot:", random.choice(responses[user]))
        break
    # Questions
    elif user in questions:
        print("Chatbot:", questions[user])
    # Unknown input
    else:
        print("Chatbot: I'm not sure. Can you rephrase?")
```
