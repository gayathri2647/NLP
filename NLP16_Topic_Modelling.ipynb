```python
# Creating example documents
doc_1 = "NLP is an artificial intelligence technology that allows computers to understand, interpret, and generate human language, enabling applications like voice assistants, email spam filters, and machine translation."

doc_2 = "Machine learning has the potential to quickly identify trends and patterns in large amounts of data that would be time consuming for humans. These insights can equip businesses, researchers, and society as a whole with new knowledge that has the potential to help them achieve their overall goals."

doc_3 = "AI is technology that enables computers and machines to simulate human learning, comprehension, problem solving, decision making, creativity and autonomy."

# Create corpus
corpus = [doc_1, doc_2, doc_3]
```

```python
# Import libraries
import string
import nltk
nltk.download('stopwords')
nltk.download('wordnet')
nltk.download('omw-1.4')  # Open Multilingual Wordnet. Used for lemmatization
from nltk.corpus import stopwords
from nltk.stem.wordnet import WordNetLemmatizer
```

```python
# remove stopwords, punctuation, and normalize the corpus
stop = set(stopwords.words('english'))
exclude = set(string.punctuation)
lemma = WordNetLemmatizer()
print(exclude)
```

```python
def clean(doc):
    stop_free = " ".join([i for i in doc.lower().split() if i not in stop])
    print(f"After removing stopwords: \n{stop_free}")
    punc_free = "".join(ch for ch in stop_free if ch not in exclude)
    print(f"After removing punctuation: \n{punc_free}")
    normalized = " ".join(lemma.lemmatize(word) for word in punc_free.split())
    print(f"After normalization: \n{normalized}")
    return normalized
```

```python
clean_corpus = [clean(doc) for doc in corpus]
```

```python
from gensim import corpora
tokenized_clean_corpus = [doc.split() for doc in clean_corpus]
dictionary = corpora.Dictionary(tokenized_clean_corpus)
doc_term_matrix = [dictionary.doc2bow(doc) for doc in tokenized_clean_corpus]
print(doc_term_matrix)
```

```python
from gensim.models import LdaModel
lda = LdaModel(doc_term_matrix, num_topics=3, id2word = dictionary)
print(lda)
print(lda.print_topics(num_topics=3, num_words=3))
```
