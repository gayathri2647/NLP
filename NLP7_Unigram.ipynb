```python
# Import libraries
import re
from collections import Counter
```

```python
# Sample text
text = """
The quick brown fox jumps over the lazy dog. The dog was not amused. The fox ran away.
"""
```

```python
# Tokenize (convert to lowercase and extract words)
def tokenize(text):
    return re.findall(r'\b\w+\b', text.lower())
```

```python
# Build Unigram Model (count word frequencies)
def build_unigram_model(text):
    tokens = tokenize(text)
    total_tokens = len(tokens)
    freq_dict = Counter(tokens)
    # Calculate probabilities for each word
    unigram_model = {word: count / total_tokens for word, count in freq_dict.items()}
    return unigram_model, total_tokens, freq_dict
```

```python
# Build model
model, total_count, frequency = build_unigram_model(text)
```

```python
# Display results
print("Total words:", total_count)
print("\nWord Frequencies:")
for word, count in frequency.items():
    print(f"{word}: {count}")

print("\nUnigram Probabilities:")
for word, prob in model.items():
    print(f"P({word}) = {prob:.4f}")
```
