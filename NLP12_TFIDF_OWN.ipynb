```python
# TF-IDF without libraries
import math
```

```python
# Define corpus (3 documents)
documents = [
    "i love nlp",
    "i love coding",
    "nlp is fun"
]
```

```python
# Tokenization
tokens_list = [doc.lower().split() for doc in documents]
```

```python
# Build vocabulary
vocab = []
for tokens in tokens_list:
    for word in tokens:
        if word not in vocab:
            vocab.append(word)
print("Vocabulary:", vocab)
```

```python
# Term Frequency (TF)
tf = []

for tokens in tokens_list:
    tf_doc = {}
    total_words = len(tokens)

    for word in vocab:
        count = tokens.count(word)
        tf_doc[word] = count / total_words

    tf.append(tf_doc)
print("\nTerm Frequency (TF):")
for i, doc_tf in enumerate(tf):
    print(f"Doc {i+1}:", doc_tf)
```

```python
N = len(documents)
df = {}
for word in vocab:
    doc_count = 0
    for tokens in tokens_list:
        if word in tokens:
            doc_count += 1
    df[word] = doc_count

# Inverse Document Frequency (IDF)
idf = {}

for word in vocab:
    idf[word] = math.log(N / df[word])

print("\nInverse Document Frequency (IDF):")
for k, v in idf.items():
    print(k, ":", round(v, 3))
```

```python
# TF-IDF Calculation
tf_idf = []

for doc_tf in tf:
    tfidf_doc = {}
    for word in vocab:
        tfidf_doc[word] = doc_tf[word] * idf[word]
    tf_idf.append(tfidf_doc)

print("\nTF-IDF Values:")
for i, doc in enumerate(tf_idf):
    print(f"Doc {i+1}:", {k: round(v, 3) for k, v in doc.items()})
```
