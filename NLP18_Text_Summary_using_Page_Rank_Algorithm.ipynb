```python
# Import libraries
import nltk
import numpy as np
import networkx as nx
from sklearn.metrics.pairwise import cosine_similarity
from sklearn.feature_extraction.text import TfidfVectorizer
from nltk.tokenize import sent_tokenize
# Download NLTK resources
nltk.download('punkt_tab')
```

```python
# Sample texts
texts =  """Natural Language Processing (NLP) helps computers understand human language.
    NLP is used for text analysis, language understanding, and communication.
    Machine learning allows computers to learn from data.
    NLP applications often use machine learning algorithms for text classification.
    Text classification assigns documents to different categories.
    Machine learning models can classify emails, reviews, and news articles.
    Sentiment analysis identifies opinions and emotions in text.
    It is commonly used to analyze customer reviews and social media posts.
    Chatbots use natural language processing to communicate with users.
    Modern chatbots use machine learning to understand questions and generate responses.
    Search engines analyze web pages and rank important information.
    PageRank is an algorithm used to determine the importance of web pages.
    PageRank ranks web pages based on their links and importance.
    A page linked by many important pages receives a higher PageRank score.
    Information retrieval finds relevant information from large collections of documents.
    Search engines use ranking algorithms to retrieve useful results.
    Text summarization produces a shorter version of a document.
    NLP techniques can identify important sentences for summarization.
    NLP is used in chatbots, sentiment analysis, text classification,
    summarization, and information retrieval.
    These applications help computers process human language."""
```

```python
# Split text into sentences
sentences = sent_tokenize(texts)
```

```python
# Convert sentences to TF-IDF vectors
vectorizer = TfidfVectorizer()
sentence_vectors = vectorizer.fit_transform(sentences)
```

```python
# Compute similarity matrix
similarity_matrix = cosine_similarity(sentence_vectors)
```

```python
# Build graph and apply PageRank
nx_graph = nx.from_numpy_array(similarity_matrix)
scores = nx.pagerank(nx_graph)
```

```python
print("Scores of Page Rank algorithm")
print(scores)
```

```python
# Rank sentences and select top ones
ranked_sentences = sorted(((scores[i], s) for i, s in enumerate(sentences)), reverse=True)
print(ranked_sentences)
```

```python
# Generate summary (top 2 sentences)
summary = " ".join([ranked_sentences[i][1] for i in range(2)])
print(summary)
```
