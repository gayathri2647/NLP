```python
# Import libraries
import nltk
from nltk.tokenize import word_tokenize, sent_tokenize
```

```python
nltk.download('punkt_tab')
```

```python
# Sample text
text = "My Family is the heart of a home. Providing a safe haven of unconditional love, support and belonging."
```

```python
# Split into sentences
sentences = sent_tokenize(text)
print(sentences)
```

```python
print("Sentences:")
# enumerate() adds a counter (index) to an iterable (like a list).
for i, sentence in enumerate(sentences, 1):
    print(f"{i}: {sentence}")
```

```python
# Split into words
words = word_tokenize(text)
print("\nWords:")
print(words)
```
