```python
# Import Libraries
from sklearn.feature_extraction.text import CountVectorizer
```

```python
# Sample documents
documents = [
    "I love machine learning",
    "Machine learning is fun",
    "I love learning new things",
    "Learning machine models is interesting",
    "I enjoy learning and building models"
]
```

```python
# Initialize CountVectorizer
vectorizer = CountVectorizer()
```

```python
# Fit and transform the documents
X = vectorizer.fit_transform(documents)
print(X)
```

```python
# Change it into dense matrix and display the answer.
print(vectorizer.vocabulary_)
```

```python
# Get the feature (word) names
feature_names = vectorizer.get_feature_names_out()
```

```python
# Convert the result to array
bow_matrix = X.toarray()
```

```python
# Display results
print("Vocabulary / Features:")
print(feature_names)
```

```python
print("\nBag of Words Matrix:")
for i, row in enumerate(bow_matrix):
    print(f"Document {i+1}: {row}")
```
