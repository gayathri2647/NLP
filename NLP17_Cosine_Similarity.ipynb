```python
# Import libraries
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.metrics.pairwise import cosine_similarity
```

```python
# Sample documents
documents = [
    "Natural Language Processing makes computers understand human language",
    "Deep Learning is used in Natural Language Processing",
    "I love playing football in the evening"
]
```

```python
# TF-IDF representation
vectorizer = TfidfVectorizer()
tfidf_matrix = vectorizer.fit_transform(documents)
```

```python
# Cosine similarity matrix
cosine_sim = cosine_similarity(tfidf_matrix, tfidf_matrix)
```

```python
# Plot heatmap
plt.figure(figsize=(6,5))
sns.heatmap(cosine_sim, annot=True, cmap="Blues", xticklabels=["Doc1","Doc2","Doc3"], yticklabels=["Doc1","Doc2","Doc3"])
plt.title("Cosine Similarity Heatmap")
plt.show()
```
