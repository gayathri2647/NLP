```python
# Topic Modelling using LDA
import string
import nltk
nltk.download('stopwords')
nltk.download('wordnet')
from nltk.corpus import stopwords
from nltk.stem import WordNetLemmatizer
!pip install gensim
import gensim
from gensim import corpora
from gensim.models import LdaModel
```

```python
# Create documents
doc1 = "Students learn online education using computers and internet."
doc2 = "Machine learning uses data to make predictions and solve problems."
doc3 = "Doctors use medical technology to diagnose patients and diseases."
```

```python
# Create corpus
corpus = [doc1, doc2, doc3]
```

```python
# Preprocessing
stop = set(stopwords.words('english'))
lemma = WordNetLemmatizer()
punctuation = set(string.punctuation)
```

```python
def clean(doc):
    words = doc.lower().split()
    words = [word for word in words if word not in stop]
    words = [word for word in words if word not in punctuation]
    words = [lemma.lemmatize(word) for word in words]
    return words
```

```python
# Clean corpus
clean_corpus = [clean(doc) for doc in corpus]
print("Clean Corpus:")
print(clean_corpus)
```

```python
# Create dictionary
dictionary = corpora.Dictionary(clean_corpus)
print("\nDictionary:")
print(dictionary.token2id)
```

```python
# Create Document-Term Matrix
doc_term_matrix = [dictionary.doc2bow(doc) for doc in clean_corpus]
print("\nDocument Term Matrix:")
print(doc_term_matrix)
```

```python
# Create LDA model
lda = LdaModel(
    doc_term_matrix,
    num_topics=3,
    id2word=dictionary,
    passes=10
)
```

```python
# Display topics
print("\nTopics:")
topics = lda.print_topics(
    num_topics=3,
    num_words=3
)
for topic in topics:
    print(topic)
```
