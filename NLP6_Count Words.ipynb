```python
# Import libraries
import re
from collections import Counter
```

```python
# Sample text
text = """
I Love Natural Language Processing Today.
"""
```

```python
# Convert text to lowercase and tokenize (split into words)
tokens = re.findall(r'\b\w+\b', text.lower())
print(tokens)
```

```python
# Count total number of words (tokens)
total_tokens = len(tokens)
print(total_tokens)
```

```python
# Count number of unique words (types)
unique_tokens = len(set(tokens))
print(unique_tokens)
```

```python
# Count occurrences of the word "i"
word_count = tokens.count('i')
print(word_count)
```

```python
# Calculate percentage of "i" in all tokens
percent = (word_count / total_tokens) * 100
print(percent)
```
