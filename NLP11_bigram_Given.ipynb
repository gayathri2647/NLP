```python
# Import libraries
import nltk
from nltk.util import bigrams
from nltk.tokenize import word_tokenize
from nltk import FreqDist
# Download tokenizer (only once)
nltk.download('punkt_tab')
```

```python
# Sample text
text = "I love NLP and I enjoy learning it."
```

```python
# Tokenize the text
tokens = word_tokenize(text.lower())
```

```python
# Generate bigrams
bigram_list = list(bigrams(tokens))
# bigrams(tokens) returns an iterator of 2-word tuples
print(bigrams(tokens))
print("Bigrams:")
for pair in bigram_list:
    print(pair)
```

```python
# Frequency Distribution of Bigrams
bigram_freq = FreqDist(bigram_list)
print("\nBigram Frequencies:")
for bigram, freq in bigram_freq.items():
    print(f"{bigram} : {freq}")
```
